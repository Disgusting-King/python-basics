# Использование библиотек в Python

Вам понадобилсь значение pi или квадратный корень из числа. Выводить формулу и провереять её самому - долго и лего ошибиться, а кто-то уже написал и отладил этот код за вас. Готовые наборы такого кода называются библиотеками: подключаете нужную одной строкой и пользуетесь, вместо того чтобы изобретать велосипед.

## Что такое библиотеки?

<b>Библиотека (или модуль) в Python</b> - это файл с кодом, содержащий функции, классы и переменные, которые вы можете использовать с своих программах.

## Импорт библиотек

Чтобы использовать библиотеку, нужно сначала импортировать её в свою программу. Python предлагает несколько способов импорта:

### Импорт всей библиотеки

```python
# Ипорт всей библиотеки
import math

# Использование функций через имя библиотеки
radius = 5
circle_area = math.pi * radius ** 2
print(f'Площадь круга радиусом {radius} равна {circle_area:.2f}')
```

### Импорт конкретных элементов

```python
# Импорт только определенных функций из модуля
from math import sqrt, floor

# Используем импортированные функции напрямую
x = 16
result = sqrt(x)
print(f'Квадратный корень из {x} равен {result}')

y = floor(3.7)
print(f'Округление 3.7 вниз: {y}')
```

### Импорт с переименованием

```python
# Импорт модуля с альтернативным именем (псевдоним)
import math as m

angle = 45
# Используем m вместо math
sin_value = m.sin(m.radians(angle))
print(f'Синус {angle} градусов равен {sin_value:.4f}')
```

## Виды библиотек в Python

В Python существует три основных вида библиотек:
- <b>Встроенные модули</b> - модули, которые уже включены в стандартную библиотеку Python и доступны сразу после установки Python.
- <b>Сторонние библиотеки</b> - модули, созданные другими разработчиками, которые нужно установить дополнительно.
- <b>Собственные модули</b> - модули, которые вы создаете сами для организации вашего кода.

### Примеры встроенных модулей

```python
# Модуль random для генерации случайных чисел
import random

# Генерация случайного числа от 1 до 10
random_number = random.randint(1, 10)
print(f'Случайное число: {random_number}')

# Модуль datetime для работы с датами и временем
import datetime

# Получение текущей даты и времени
current_date = datetime.datetime.now()
print(f'Текущая дата и время: {current_date}')
```

## Поиск функций и документации

При работе с библиотеками важно знать, как найти информацию о доступных функциях. Python предоставляет несколько способов:

### Использование функции help()

```python
# Получение информации о модуле
import math
help(math)
```

### Использование функции dir()

```python
# Получение списка всех доступных атрибутов и методов
import random
attributes = dir(random)
print(attributes)
```
