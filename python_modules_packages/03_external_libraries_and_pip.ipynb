# Сторонние библиотеки и pip

Понадобилось обратиться к чужому API, разобрать веб-страницу или проанализировать таблицу на миллион строк. В стандартной библиотеке кое-что для этого есть, но сообщество написало инструменты намного удобнее: <b>requests, beatutifulsoup4, pandas</b> и тысячи других. Это и есть сторонние библиотеки - то, что не входит в поставку Python и ставится отдельно через <b>pip</b>, стандартный менеджер пакетов.

## Что такое сторонние библиотеки?

<b>Сторонние библиотеки</b> - это модули Python, которые не входят в стандартную библиотеку и разрабатываются независимыми разработчиками или организациями.

## Что такое pip?

<b>pip (Package Installer for Python)</b> - это система управления пакетами, которая используется для установки и управления программными пакетами, написанными на Python. Он скачивает пакеты из <b>Python Package Index (PyPI)</b>, а также умеет обновлять, удалять и фиксировать версии зависимостей проекта.

## Установка и использование pip

Обычно pip уже установлен вместе с Python. Проверить наличие pip можно так:

```python
pip --version
```

### Основные команды pip

#### Установка пакета

```python
# pip install requests
```

#### Установка определенной версии

```python
# pip install requests==2.25.1
```

#### Список установленных пакетов

```python
pip list
```

#### Удаление пакета

```python
# pip uninstall requests -y
```

## Виртуальные окружения

Если ставить все библиотеки в системный Python, рано или поздно вы упретесь в конфликт: один проект требует <b>django 3.0</b>, а другой - <b>django 4.2</b>, а версия может быть только одна. <b>Стандартная практика - заводить отдельное виртуальное окружение для каждого проекта.</b>

<b>Вирутальное окружение (venv)</b> - это изолированная среда Python, в которой можно устанавливать пакеты, не влияя на другие проекты или системный Python.

### Создание виртуального окружения

```python
# Создание вируального окружения
# python -m venv myenv

# Активация виртуального окружения
# В Windows:
# myenv/Scripts/activate

# В macOS/Linux:
# source myenv/bin/activate

# После активации в начале командной стркои появится имя окружения
# (myenv) $
```

### Установка пакетов в виртуальное окружение

```python
# Устанавливаем пакеты в активированное виртуальное окружение
# pip install pandas matplotlib
```

### Сохранение и установка зависимостей

```python
# Сохранение списка установленных пакетов
# pip freeze > requirements.txt

# Содержимое файла requirement.txt будет выглядеть примерно так:
# matplotlib==3.7.2
# numpy==1.25.2
# pandas==2.0.3
# ...

# Установка пакетов из файла requirements.txt
# pip install -r requirements.txt
```
