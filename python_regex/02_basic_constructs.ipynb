# Базовые конструкции

## Литералы: Просто текст

Добро пожаловать во второй модуль! Здесь мы начнем строить сами шаблоны. 

Самый простой тип реглярного выражения - это просто <b>литерал</b> (literal). Это умное слово означает простую вещь: <b>символ, который обозначает сам себя</b>.

Если вы напишите регулярку - cat, движок будет искать:
1. Символ c, за которым сразу следует...
2. Символ a, за которым сразу следует...
3. Символ t.

В этом плане регулярные выражения работают точно так же, как обычный поиск в Word или Блокноте. Большинство букв и циффр являются литералами:
- Шаблон 2023 найдет строку "2023".
- Шаблон Привет найдет строку "Привет".

### Регистрозависимость

Самое важное правило при работе с литералами: <b>Регулярные выражения чувствительны к регистру (Case Sensitive)</b>.

Для компьтера бука "А" и бука "а" - это совершенно разные символы (у них разные коды в таблице ASCII/Unicode). Поэтому, если вы напишете шаблон с большой буквы, а в тексте она маленькая - магии не случится.

```python
import re

text = 'Я люблю Python'
print(re.search(r'Python', text))
print(re.search(r'python', text))
```

## Метасимвол "Точка"

Если бы регулярные выражения состояли только из литералов (как cat), они были бы бесполезны. Вся сила RegEx кроется в <b>метасимволах</b> - специальных символах, которые обладают "сверхспособностями".

В регулярных выражениях точка означает: <b>Один любой символ. КРОМЕ! символа переноса строки (\n)</b>. Исторически так сложилось, потому что утилиты для обработки текста часто читали файлы построчно. Перенос строки считался разделителем, за который нельзя заходить.

### Пример

```python
import re

pattern = r'a.c'
print(re.match(pattern, 'abc'))
print(re.match(pattern, 'a\nc'))
```

## Экранирование

Мы выяснили, что символы вроде ".", "?", "(" - это <b>метасимволы</b>. Они не просто текст, они выполняют команды.

Но что, если наша задача - найти в тексте <b>настоящую точку</b>? Например, мы ищем имя файла image.png или IP-адрес 192.168.0.1.

Если мы напишем шаблон image.png, точка сработает как Джокер:
- Шаблон найдет image.png (точка совпала с точкой)
- Но он также найдет imageXpng, image2png и image_png

Нам нужно как-то сказать движку: "Слушай, вот сейчас забудь про спецэффекты. Я хочу просто точку".

### Обратный слэш (\\)

Чтобы превратить <b>метасимвол</b> обратно в <b>литерал</b>, перед ним нужно поставить обратный слэш \\. Этот процесс называется <b>экранированием</b> (escaping).

Слэш работает как "глушилка" магии для следующего за ним символа.
- <b>.</b> - любой символ
- <b>\\.</b> - просто точка
- <b>?</b> - квантификатор (0 или 1 раз)
- <b>\\?</b> - просто вопросительный знак
- <b>(</b> - начало группы
- <b>\\(</b> - просто круглая скобка

### Пример

```python
import re

text1 = 'google.com'
text2 = 'googleXcom'

pattern_bad = r'google.com'

print(re.match(pattern_bad, text1))
print(re.match(pattern_bad, text2))

pattern_good = r'google\.com'

print(re.match(pattern_good, text1))
print(re.match(pattern_good, text2))
```

## Квадратные скобки: Строгий отбор

Представьте, что вы ищете слово, которое может ьть написано двумя способами: <b>gray</b> (американский) и <b>grey</b> (британский).

### Наборы символов (Character Sets)

Для этого используются квадратные скобки []. Всё, что ввы пречислите внутри них, работает по логике <b>ИЛИ</b>.

Шаблон [abc] означает: "ОДИН символ, который является 'a' ИЛИ 'b' ИЛИ 'c'"

Вернемся к примеру с цветом. Правильный шаблон: 'gr[ae]y'.

### Спецсимволы внутри скобок теряют силу

У квадратных скобок есть приятный бонус. Внутри них большинство спецсимволов перестают быть специальными и становятся обычными текстовыми символами.

Если вы хотите найти точку или вопросительный знак, внутри скобок их можно (чаще всего) не экранировать.
- <b>[.]</b> - ищет реальную точку
- <b>[?]</b> - ищет реальный знак вопроса
- <b>[(]</b> - ищет реальную скобку

## Диапазоны: От А до Я

Перечислять символы вручную - это долго. Если вы хотите найти любую цифру, писать [0123456789] утомительно. А если нужно найти любую букву английского алфавита? Писать все 26 символов?

### Дефис-спаситель

Внутри квадратных скобок дефис - становится метасимволом, который создает диапазон.
- <b>[0-9]</b> - любая цифра от 0 до 9
- <b>[a-z]</b> - любая строчная латинская буква
- <b>[A-Z]</b> - любая заглавная латинская буква

Вы <b>можете комбинировать диапазоны</b> в одних скобках. Например: <b>[a-zA-Z0-9]</b>

### А если мне нужен сам дефис?

Если дефис - это спецсимвол диаапазона, как найти в тексте знак минуса? Есть два способа:
1. Поставить его в самом начале или в самом конце внутри скобок. Там он не может создавать диапазон.
    - [-0-9] - дефис ИЛИ цифры
    - [a-z-] - буквы ИЛИ дефис
2. Экранировать его: [/-]

## Исключающие наборы: "Всё, кроме..."

Иногда проще сказать, чего мы не хотим видеть, чем перечислять всё, что хотим. Представьте, что м ищем в тексте ошибки, где вместо цифр кто-то ввел буквы. Нам нужно найти: <b>"Любой символ, который НЕ является цифрой."</b>

Для этого в регулярных выражениях используется символ <b>карет</b> (крышечка) - ^

### Синтаксис отрицания

Чтобы превратить набор в исключающий, поставьте <b>"^"</b> сразу после открывающей скобки.
- <b>[abc]</b> - найти a, b и c
- <b>[^abc]</b> - найти любой символ, кроме a, b и c

## Шорткаты: "Большая тройка"

В программировании ценится краткость. Писать каждый раз [0-9] или [a-zA-Z0-9] - это долго и портит читаемость кода.

Для самых популярных наборов символов сохданы <b>специальные сокращения (shortcuts)</b>. Они состоят из обратного слэша и маленькой буквы.

Их всего три, но они покроют 90% ваших потребностей.
1. \d - <b>Digit</b> (Цифра). Аналог: [0-9]
2. \s - <b>Space</b> (Пробел). Аналог: [\t\n\r\f\v]
3. \w - <b>Word</b> (Латинские буквы и цифры + "_"). Аналог: [_a-zA-Z0-9]


## Обратные шорткаты

В регулярных выражениях царит гармония. Для каждого шортката, который что-то ищет, существует его злой двойник, который это исключает.

Правило запоминания элементарное:
- <b>Маленькая буква</b> (\d, \s, \w) - ищет совпадение
- <b>Большая буква</b> (\D, \S, \W) - работает как НЕ

Это то же самое, что использовать исключающий набор [^...], но гораздо короче.

1. <b>\D</b> - НЕ цифра (Not Digit). Аналог: [^0-9]
2. <b>\S</b> - НЕ пробел (Not Space). Аналог: [^\t\n\r]
3. <b>\W</b> - НЕ слово (Not Word). Аналог: [^a-zA-Z0-9_]
