# Группировка

Мы переходим к одному из самых мощных разделов - <b>Группировке</b>. До этого момента наши квантификаторы (+, *, ?) были "одиночками". Они влияли только на тот символ, который стоял непосредственно слева от них. 

А что, если мы хотим повторить какую-било последовательность целиком? Нам нужно "склеить" символы в один блок. Для этого используются <b>Круглые скобки</b>.

### Логическое объединение

Если мы напишем (abc)+, движок регулярных выражений поймет это так: "Возьми блок 'abc' и повтори этот блок 1 или много раз"

### Применение с другими квантификаторами

Круглые скобки позволяют применять любую логику к целым фразам.
1. <b>Необязательная фраза</b>: (text )?file
2. <b>Выбор из вариантов</b>: (Monday|Tuesday|Wednesday)

## Нумерация групп

Круглые скобки () выполняют в регулярных выражениях двойную работу:
1. <b>Группировка</b> (как мы видели в прошлом шаге) - позволяют применить квантификатор к целому блоку.
2. <b>Захват</b> (Capturing) - это самое интересное. Всё, что попало внуть скобок, запоминается движком в отдельную "ячейку памяти".

Это позволяет нам извлекать конкретные кусочки данных из строки.

### Нулевая группа

Когда re.search() находит совпадение, он возвращает объект Match. Мы уже использовали метод .group() без аргументов. На самом деле, по умолчанию вызывается .group(0).

<b>Группа 0</b> - это всё совпадение целиком. Весь текст, который подошел под ваш шаблон.

### Вложенные группы

Сложности начинаются, когда одни скобки находятся внутри других. Как нумеруются группы в шаблоне вроде: (Я (люблю (Python)))?

<b>Правило железное и простое</b>: Нумерация идет по ОТКРЫВАЮЩИМ скобкам слева направо.

<b>Разбор шаблона: (Model: (\d+))</b>

Представим, что мы ищем запись вида Model: 800. Мы хотим захватить и всю фразу целиком в переменную, и отдельно номер модели.
- <b>Группа 1</b>: (Model: (\d+)) - Внешняя скобка. Она открывается первой.
- <b>Группа 2</b>: (\d+) - Внутренняя скобка. Она открывается второй.

```python
import re

text = "Specification: Model: 800"
pattern = r"(Model: (\d+))"

match = re.search(pattern, text)
print(f"Группа 1: {match.group(1)}") # Model: 800
print(f"Группа 2: {match.group(2)}") # 800
```

### Метод .groups()

Если групп много, доставать их по одной утомительно. У объекта Match есть метод <b>.groups()</b>. Он возвращает <b>кортеж</b> всех захваченных подгрупп.

```python
# Дата: День-Месяц-Год
pattern = r"(\d{2})-(\d{2})-(\d{4})"
match = re.search(pattern, "05-11-2023")

print(match.groups())
# Вывод: ('05', '11', '2023')

# Удобная распаковка в Python:
day, month, year = match.groups()
print(f"Год рождения: {year}")
```

## Задачи

<b>Условие</b>:

Нам нужно распарсить дату в формате YYYY-MM-DD (например, 2023-05-12).
Напишите регулярное выражение с тремя группами захвата:

1. Год (4 цифры).
2. Месяц (2 цифры).
3. День (2 цифры).

Выведите год, месяц и день на отдельных строках.

<b>Формат ввода</b>:
Строка с датой.

<b>Формат вывода</b>:
Три строки:
Year: ...
Month: ...
Day: ...

Если формат неверный, выведите Wrong format.

```python
import re

pattern = r'(\d{4})-(\d{2})-(\d{2})'
text = input()
match = re.search(pattern, text)
rows = ['Year', 'Month', 'Day']

if match:
    for row, value in zip(rows, match.groups()):
        print(f'{row}: {value}')
else:
    print('Wrong format')
```

### Задача 4. Вложенные группы

<b>Условие</b>:

У нас есть данные о точках на карте в формате Point(10,20).
Нам нужно достать:
1. Всю координатную часть целиком: 10,20.
2. Отдельно X: 10.
3. Отдельно Y: 20.

Напишите шаблон с вложенными группами.
Внешняя группа захватывает цифры,цифры.
Внутренние группы захватывают числа по отдельности.

<b>Формат ввода</b>:
Строка.

<b>Формат вывода</b>:
Строка вида: Full: 10,20 | X: 10 | Y: 20.

```python
import re

pattern = re.compile(r'Point\(((\d{2}),\s*(\d{2}))\)')
text = input()

for match in pattern.finditer(text):
    print(f'Full: {match.group(1)} | X: {match.group(2)} | Y: {match.group(3)}')
```

## Именованные группы

Нумерованные группы group(1), group(2) - это удобно, пока их мало. Но представьте сложный шаблон с 10 группами. В коде вы пишете match.group(7).
- Что там лежит? Год? Имя? IP-адрес?
- А если вы завтра добавите новую группу в начало шаблона, то 7-я группа станет 8-й, и вам придется переписывать весь код.

Чтобы избежать путаницы, Python позволяет давать группам имена.

### Синтаксис

Синтаксис выглядит немного пугающе, но к нему быстро привыкаешь. Внутри открывающей скобки нужно добавить специальную конструкцию ?P<имя>.
- Было: (\d{4})
- Стало: (?P<_year>\d{4})

### Доступ по имени

Теперь, когда у группы есть имя, мы можем обращаться к ней как к ключу словаря.

```python
# match.group('year')
```

Это делает код читаемым. Сравните:
- print(match.group(4)) - Напечатай четвертую штуковину
- print(match.group('status_code')) - Напечатай код статуса

### Пример: Парсинг сайта

Доступ, у нас есть имя файла с фильмом: "Matrix_1999_1080p.mkv". Нам нужно вытащить название, год и качество.

```python
import re

filename = "Matrix_1999_1080p.mkv"

# Составляем шаблон с именами
# 1. title: буквы/цифры
# 2. year: 4 цифры
# 3. resolution: цифры + буква 'p'
pattern = r"(?P<title>\w+)_(?P<year>\d{4})_(?P<resolution>\d+p)\.mkv"

match = re.match(pattern, filename)

if match:
    # Обращаемся по именам
    print(f"Фильм: {match.group('title')}")
    print(f"Год:   {match.group('year')}")
    
    # Обратите внимание: доступ по номеру ВСЁ ЕЩЁ РАБОТАЕТ
    # Группа 'title' всё ещё является группой №1
    print(f"Через индекс: {match.group(1)}")

# Вывод:
# Фильм: Matrix
# Год:   1999
# Через индекс: Matrix
```

### Метод .groupdict()

Для именованных групп есть еще один супер-полезный метод - .groupdict(). Он возвращает словарь Python, где ключи - это имена ваших групп, а значения - найденный текст.

```python
data = match.groupdict()
print(data)
```

Это идеально для конвертации данных в JSON или сохранения в базу данных. Вам не нужно вручную собирать словарь, re делает это за вас.

## Обратные ссылки: Эффект Эха

До сих пор наши шаблоны были статичными. Если мы писали \d\d, мы искали любую цифрку, а потом любую цифру. Это найдет "12", "55" и "90".

Но что, если мы хотим найти <b>повторы</b>? Например, мы ищем опечатки в тексте: "Я пошел в в магазин". Мы не знаем, какое слово будет продублировано. Это может быть "в в", "на на" или "кот кот".

Здесь нам нужны <b>Обратные ссылки (Backreferences)</b>

### Что это такое?

Обратная ссылка говорит движку: "Найди здесь ТО ЖЕ САМОЕ, что ты уже нашел и положил в Группу N". Это не просто повторение правила (как \w), это повторение конкретного захваченного текста.

### 1. Нумерованные ссылки \1, \2

Если у вас есть группа захвата (...) под номером 1, то в том же самом шаблоне (чуть дальше) вы можете сослаться на её содержимое с помощью \1.

<b>Шаблон</b>: \b(\w+)\s+\1
- \b - граница слова (чтобы не находить "one" внутри "stone one")
- (\w+) - находит слово и запоминает его в Группу 1 (например, 'the')
- \s+ - один или несколько пробелов
- \1 - проверяет, что следующее слово в точности равно содержимому Группы 1 ('the')

#### Пример: Ловля опечаток

```python
import re

text = "Я пошел в в магазин. Там было очень очень людно. А еще стон тон."

# Шаблон: Граница слова, Слово, пробел, то же самое слово
# Мы добавили \b, чтобы избежать частичных совпадений (как в "стон тон")
pattern = r"\b(\w+)\s+\1"

duplicates = re.findall(pattern, text)

print(duplicates)
# Вывод: ['в', 'очень']
# "стон тон" проигнорирован, так как мы требуем начало слова (\b)
```

Если бы мы написали (\w+)\s+(\w+), мы бы нашли просто два любых слова подряд.

### 2. Именованные ссылки (?P=name)

Если вы используете имменованные группы (что мы горячо рекомендуем), ссылаться на них по номеру \1 можно, но некрасиво и ненаглядно. Для этого есть специальный синтаксис ссылки по имени: (?P=name)

#### Пример: Валидация HTML (по-простому)

Нам нужно найти текст внутри тегов. Но открывающий и закрывающий теги должны совпадать!

```python
import re

html = "Правильно: <b>Жирный</b>. Ошибка: <i>Курсив</b>."

# 1. (?P<tag>\w+) - Запоминаем имя тега (b, i, div...) в группу 'tag'
# 2. .*?          - Текст внутри
# 3. (?P=tag)     - Ожидаем закрывающий тег с ТЕМ ЖЕ именем
pattern = r"<(?P<tag>\w+)>(.*?)</(?P=tag)>"

for match in re.finditer(pattern, html):
    print(f"Нашел корректный блок: {match.group(0)}")

# Вывод:
# Нашел корректный блок: <b>Жирный</b>
```

Второй кусок (<_i>...</_b>) был проигнорирован, потому что содержимое группы tag стало равно "i", и движок ожидал закрывающего </_i>, а встретил </_b>.

## Рефакторинг текста: Ссылки при замене

В прошлом шаге мы использовали \1 внутри шаблона поиска, чтобы найти повторы. Но настоящая магия обратных ссылок раскрывается в функции <b>re.sub()</b>.

Представьте задачы: у вас есть даты в формате <b>YYYY-MM-DD</b> (2023-12-31), а вам нужно переделать их в наш привычный <b>DD.MM.YYYY</b> (31.12.2023).

Вы не можете просто заменить всё на фиксированную строку. Вам нужно взять кусочки найденного текста и <b>переставить их местами</b>.

### Как это работает?

В функцию re.sub в качестве замены м передаем строку, содержащую ссылки на группы.
1. Классический способ: \1, \2...
2. Безопасный способ (Python): \g<1>, \g<2>...
3. Именованный способ: \g<_name>

### Пример 1. Перестановка даты

```python
import re

text = "Срок действия: 2023-10-05"

# Шаблон с 3 группами: (Год)-(Месяц)-(День)
pattern = r"(\d{4})-(\d{2})-(\d{2})"

# Строка замены: "Группа3.Группа2.Группа1"
replacement = r"\3.\2.\1"

result = re.sub(pattern, replacement, text)
print(result)
# Вывод: Срок действия: 05.10.2023
```

Движок взял содержимое 3-й скобки (05), поставил точку, взял 2-ю (10), точку, и 1-ю (2023). Готово!

### Проблема неоднозначности \1

Почему в Python придумали странный синтаксис \g<1>? Потому что \1 может вести себя непредсказуемо.

Представьте, что вы хотите заменить найденную группу №1 и сразу приписать к ней цифру 0. Вы пишете r'\10'

Движок в замешательстве:
- Это ссылка на <b>Группу №1</b>, за которой идет ноль?
- Или это ссылка на <b>Группу №10</b>?

Чтобы избежать этой каши, используйте <b>явные границы: \g<1>0</b>

### Пример 2. Именованные группы

Если вы используете именованные группы (?P<_name>...), ссылаться на них при замене нужно через \g<_name>. Это делает код замены невероятно понятным. Давайте поменяем формат имени с "Фамилия,Имя" на "Имя,Фамилия".

```python
import re

text = "Список: Иванов, Иван; Петров, Петр"

# Шаблон: (Фамилия), пробел (Имя)
pattern = r"(?P<last>\w+), (?P<first>\w+)"

# Замена: <Имя> <Фамилия>
# Мы просто меняем местами блоки по их именам
replacement = r"\g<first> \g<last>"

result = re.sub(pattern, replacement, text)

print(result)
# Вывод: Список: Иван Иванов; Петр Петров
```

### Дублирование текста

Вы можете использовать одну ссылку сколько угодно раз. Например, превратим слово в HTML-тег: Title -> <_h1>Title</_h1>

```python
text = "Chapter 1"
# Оборачиваем всё совпадение (\g<0>) в теги
print(re.sub(r"Chapter \d", r"<h1>\g<0></h1>", text))

# Вывод: <h1>Chapter 1</h1>
```

<b>Условие</b>:

Нам нужно извлечь информацию о фильме из строки вида Matrix (1999).

2. (?P<_year>...) — год (4 цифры) в скобках.

### Задача 3. Валидация тегов

### Задача 5. Зеркальные границы

<b>Условие</b>:

Пример: aba (начинается на a, заканчивается на a).

Пример: abc (начинается на a, заканчивается на c) — не подходит.

Используйте именованную группу для первой буквы и ссылку на неё в конце.

## Оператор ИЛИ (|)

В программировании мы часто используем условия: if a or b.

В регулярных выражениях тоже есть свой OR. Он обозначается вертикальной чертой "|". Его логика проста:
- cat|dog - найдет "cat" или "dog"
- reg|green|blue - найдет один из цветов

Казалось бы всё просто, но здесь кроется подвох с приоритетом.

### Глобальный разрыв

Оператор | имеет самый низкий приоритет из всех. Это значит, что он делит всё ваше регулярное выражение на две независимые части.

Представьте, что вы хотите найти музыкантов по имени Jimi: Jimi Hendrix или Jimi Page. 

<b>Плохой шаблон</b>

Jimi Hendrix|Page

Как это видит движок:
1. Либо полная фраза Jimi Hendrix
2. Либо просто слово Page (без имени Jimi)

<b>Хороший шаблон</b>

Чтобы ограничить действие "трубы", мы должны использовать круглые скобки.

Jimi (Hendrix|Page)

Теперь движок видит:
1. Слово Jimi
2. Затем выбор: или Heendrix, или Page

### Порядок имеет значение

Движок проверяет варианты слева направо и останавливается, как только нашел совпадение. Это может привести к ошибкам, если одно слово является частью другого:
- <b>Пример</b>: Мы ищем слова "car" и "cart".
- <b>Шаблон</b>: car|cart
- <b>Текст</b>: "I bought a cart"

<b>Что происходит?</b>
1. Движок видит car
2. Проверяет текст cart
3. В начале слова cart есть car
4. Успех! Движок нашел car и довольный остановился. Букву t он проигнорировал.

<b>Как исправить?</b>

Всегда ставьте более длинные/специфичные варианты первыми: <b>cart|car</b>

### Практика

```python
import re

text = "Jimi Hendrix and Janis Joplin"

# 1. Без скобок (Ошибка логики)
# Мы хотели найти полные имена, но 'Joplin' оторвалась от 'Janis'
# Шаблон ищет: "Jimi Hendrix" ИЛИ просто "Janis"
print(re.findall(r"Jimi Hendrix|Janis", text))
# Вывод: ['Jimi Hendrix', 'Janis'] 
# (Обратите внимание: Joplin вообще не найдена!)

# 2. Со скобками (Правильно)
# Мы ищем: (Jimi ИЛИ Janis) + Пробел + (Hendrix ИЛИ Joplin)
pattern = r"(Jimi|Janis) (Hendrix|Joplin)"
print(re.findall(pattern, text))
# Вывод: [('Jimi', 'Hendrix'), ('Janis', 'Joplin')]
```

## Незахватывающие группы: Группировка без багажа

Мы уже знаем, что круглые скобки (...) делают две вещи одновременно:
1. Группируют символы
2. Захватывают текст в память

Но что, если нам нужно первое, но совершенно не нужно второе?

### Проблема "Лишней памяти"

Представьте, что вы ищете слова с окончанями: industr(y|ial). Вам нужно слово целиком. Вы используете скобки только для того, чтобы ограничить действие оператора ИЛИ (|).

Если вы используете обычные скобки, движок послушно сохранит "y" или "ial" в отдельную ячейку памяти (Группа 1).
- <b>Минус 1</b>: Это тратит (немного) памяти и времени.
- <b>Минус 2</b>: Это сбивает нумерацию. Если у вас в шаблоне много важных групп, лишние "технические" группы путают порядок.
- <b>Минус 3</b>: Это ломает работу re.findall()

### Решение

Чтобы отключить функцию "захвата", добавьте ?: сразу после открывающей скобки. Она работает как обычная скобка (группирует), но не создает нумерованную группу и не сохраняет текст в память.

### Пример 1. Чиним fidnall

Это самая частая причина использования (?:...) в Python. Вспомним, как работает findall:
- Если групп нет - возвращает список полных совпаденией (строки).
- Если есть группа (...) - возвращает список содержимого групп (вырезает только то, что в скобках).

<b>Ситуация</b>: Мы ищем смех hahaha.

```python
import re

text = "hahaha"

# ❌ Обычная группа (Capturing)
# Мы использовали скобки, чтобы применить плюс к "ha"
# findall видит скобки и думает: "Ага, пользователю нужно только то, что внутри!"
# Он возвращает последнее совпадение группы
print(re.findall(r"(ha)+", text))
# Вывод: ['ha'] (Это не то, что мы хотели!)

# ✅ Незахватывающая группа (Non-capturing)
# Мы говорим: "Эти скобки только для структуры. Не вырезай их."
# findall видит, что захватывающих групп нет, и возвращает строку целиком.
print(re.findall(r"(?:ha)+", text))
# Вывод: ['hahaha'] (Ура!)
```

### Пример 2. Струтура без мусора

Допустим, мы ищем имя и фамилию: Mr. Smith.

Обращение (Mr или Ms) нам нужно только для поиска, само по себе оно нам не интересно. Нам нужна фамилия.

```python
text = "Mr. Smith"

# (?:Mr|Ms) - проверяем префикс, но не занимаем под него номер группы
# (\w+)     - а вот фамилию захватываем в Группу 1
pattern = r"(?:Mr|Ms)\. (\w+)"

match = re.search(pattern, text)
print(match.group(1)) 
# Вывод: Smith
# Обратите внимание: Smith стала группой №1, потому что первая скобка была проигнорирована.
```

### Оптимизация

Хотя на современных компьютерах это почти незаметно, незахватывающие группы работают быстрее. Движку не нужно выделять память под сохранение подстроки и копировать туда данные.

В огромных текстах и сложных шаблонах это считается хорошим тоном (Best Practice).

### Задача 3. Чистота findall

<b>Условие</b>:

Это задача на понимание разницы между (...) и (?:...).
Нам нужно найти имена людей с обращениями: Mr. Smith, Ms. Stone, Mrs. Doe.
1. Если мы напишем (Mr|Ms|Mrs)\. \w+, то функция re.findall вернет только обращения (то, что в скобках): ['Mr', 'Ms'].
2. Чтобы findall вернул всё целиком (['Mr. Smith', 'Ms. Stone']), нужно использовать незахватывающую группу: (?:Mr|Ms|Mrs).

Напишите правильный шаблон и выведите список найденных имен.

<b>Формат ввода</b>:
Строка с именами.

<b>Формат вывода</b>:
Список полных имен (list).

```python
import re

pattern = r'(?:Mrs?|Ms)\. \w+'
text = input()

print(re.findall(pattern, text))
```

### Задача 5. Разные разделители

<b>Условие</b>:

У нас есть идентификаторы, которые могут начинаться либо с ID:, либо с символа #.
Примеры: ID: 123, #123.

Нам нужно извлечь только цифры с помощью re.findall.
Для этого:
1. Часть с префиксом (ID: или #) оберните в незахватывающую группу (?:...).
2. Цифры \d+ оберните в захватывающую группу (...).
Напоминание: Если в findall есть захватывающая группа, он вернет только её содержимое.

<b>Формат ввода</b>:
- Текст с ID.

<b>Формат вывода</b>:
- Список строк (только цифры).

```python
import re

pattern = r'(?:#|ID:)\s*(\d+)'
text = input()

print(re.findall(pattern, text))
```
