# Методы модуля re

## Оптом дешевле: re.findall()

Мы переходим к четвертому модулю - самому практичному. До этого мы учились искать одно совпадение. Но в реальности нам чаще нужно "вытащить все email-адреса со страницы" или "собрать все цены из прайс-листа".

Если вы будете использовать re.search() в цикле - это будет долго и больно. Знакомьтесь с главной рабочей лошадкой парсинга - re.findall()

### Как это работает?

Функция <b>re.findall(pattern, string)</b> сканирует строку слева направо и находит ВСЕ непересекающиеся совпадения.

<b>re.findall()</b> возвращает обычный список (list) строк. Никаких Match Object. Никаких <b>.group()</b>. Вы получаете чистые данные, готовые к использованию.

## findall и Группы: Внезапный поворот

Функция findall() кажется простой и понятной, пока вы не добавите в шаблон круглые скобки (). В этот момент поведение функции несколько меняется.

### Три лица findall

В зависимости от того, сколько групп захвата (пар круглых скобок) есть в вашем шаблоне, функция возвращает данные в разных форматах. Рассмотрим это на примере парсинга строки: "Имя: Alex, Имя: Bob"

```python
import re

text = 'Имя: Alex, Имя: Bob'
```

#### Нет скобок - Список строк

```python
# Если в шаблоне нет групп, findall возвращает полное совпадение.

pattern = r'Имя: \w+'
print(re.findall(pattern, text))

# Все как обычно. Мы получили список строк.
```

#### Одни скобки - Вырезка

```python
# Если в шаблоне есть одна группа (...), findall находит полное совпадение,
# но возвращает в список ТОЛЬКО то, что внутри скобок.
# 
# Это очень удобно для точечного извлечения данных (scraping). Нам не нужно
# потом обрезать лишнее "Имя: " вручную.

pattern = r'Имя: (\w+)'
print(re.findall(pattern, text))
```

#### Много скобок - Список кортежей

```python
# Если в шаблоне две и более групп, функция возвращает список кортежей.
# Каждый кортеж содержит части, попавшие в первую, вторую и т.д. группу.

text = '2023-10'
pattern = r'(\d{4})-(\d{2})'

print(re.findall(pattern, text))

# Это позволяет за один проход разбить сложную строку на составляющие.
```

### Опасная ловушка

Иногда мы используем скобки не для того, чтобы что-то извлечь, а просто чтобы применить квантификатор к целому слову. Например, мы ищем "ha" один или более раз: (ha)+

```python
import re

text='hahaha'
print(re.findall(r'(ha)+', text))
```

<b>Как это лечить?</b>

Если вам нужны скобки для логики (группировки), но вы не хотите менять поведение findall, используйте незахватывающие группы (?:...). О них подробнее в 5 модуле, но просто знайте: они чинят findall.

```python
print(re.findall(r'(?:ha)+', text))
```

## re.split()

Вы наверняка пользовались стандартным методом строк <b>text.split(',')</b>. Он хорош, но слаб. Он умеет разбивать текст только по одному фиксированному символу.

А что если данные грязные? "Яблоко, Груша; Ананас|Банан". Здесь разделители разные: запятая, точка с запятой, вретикальная черта, да еще и пробелы где-то здесь есть, а где-то нет.

Здесь на помощь приходит <b>re.split()</b>.

### Умное разделение

Функция <b>re.split(pattern, string)</b> разбивает строку каждый раз, когда находит соответствие шаблону.

Давайте разрежем нашу фруктовую строку. Шаблон разделителя: <b>[;,|]\s*</b>

```python
import re

text = 'Яблоко, Груша; Ананас|Банан'
fruits = re.split(r'[;,|]\s', text)

print(fruits)
```

Разделители исчезли, остались только чистые данные.

### Куда деваются разделители?

По умолчанию <b>re.split()</b> работает как нож: он вырезает разделитель и выбрасывает его. В результирующем списке его нет.
- Строка: 100+200
- Шаблон: \\+
- Результат: ['100', '200'] (Плюс исчез)

### Сохранение разделителя

Иногда нам нужно разбить текст, но сохранить то, по чему мы разбивали. Например, если мы пишем калькулятор, нам важно знать, что между числами был именно плюс, а не минус.

Для этого нужно просто обернуть шаблон разделителя в круглые скобки ().

Если в шаблоне есть Группа захвата, re.split() вернет и куски текста, и сами разделители.

```python
import re

expression = '100+200-50'

print(re.split(r'[\+\-]', expression))
print(re.split(r'([\+\-])', expression))
```

## re.sub()

В любом текстовом редакторе есть функция "Find & Replace". В Python у строк есть метод .replace('old', 'new').

Но что, если задача звучит так:
- Замени все даты в формате YYYY-MM-DD на [ДАТА]
- Удали все лишние пробелы между словами
- Скрой все номера банковских карт

Обычный .replace() здесь бессилен, потому что каждый раз разный. Здесь в игру вступает re.sub (от слова Substitution - подстановуа).

### Синтаксис

new_string = re.sub(pattern, replacement, string, count=0)

Аргументы:
1. <b>pattern</b>: ЧТО ищем
2. <b>replacement</b>: НА ЧТО меняем
3. <b>string</b>: ГДЕ меняем
4. <b>count</b>: СКОЛЬКО РАЗ меняем

### Пример 1. Генеральная уборка

Допустим, у нас есть текст с кучей лишних пробелов, табов и переносов строк. Мы хотим превратить его в одну аккуратную строку с одиночными пробелами.

Шаблон \s+ означает "один или более пробельных символов".

```python
import re

text = "Привет,    мир! \n  Как   твои\tдела?"

# Заменяем любую кучу пробелов на один обычный пробел " "
clean_text = re.sub(r"\s+", " ", text)

print(clean_text)
# Вывод: 'Привет, мир! Как твои дела?'
```

### Пример 2. Цензура

Спрячем номера телефонов в тексте. Допустим, формат номера прост: \d-\d{3}-\d{3} (например, 8-900-123)

```python
text = "Звоните 8-900-555 или 7-495-123 по срочным вопросам."

# Заменяем шаблон на строку "[СКРЫТО]"
censored = re.sub(r"\d-\d{3}-\d{3}", "[СКРЫТО]", text)

print(censored)
# Вывод: 'Звоните [СКРЫТО] или [СКРЫТО] по срочным вопросам.'
```

## re.sub() Level Up

В прошлом шаге мы меняли одно слово на другое. Это работает, если замена статична. Но что, если нам нужна логика? Представьте задачи:
- Найти все цены в тексте и умножить их на курс доллара
- Найти все слова и перевести их в ВЕРХНИЙ РЕГИСТР
- Найти даты и поменять формат с 2023-10-05 на 05.10.2023

Простая строка замены здесь не поможет. Она не умеет умножать числа. Здесь на помощь приходят <b>Callback-функции</b> (функции обратного вызова)

### Как это работает?

Вторым аргументом в re.sub() можно передать не строку, а имя функции. 

re.sub(pattern, function_name, text)

Алгоритм работы движка меняется:
1. Движок находит совпадение по шаблону.
2. Он НЕ заменяет текст сразу.
3. Он создает объект Match (как при search) и передает его в вашу фукцию.
4. Ваша функция делает вычисления и возвращает строку.
5. Движок вставляет эту строку в текст вместо найденного фрагмента.

### Анатомия функции-замены

```python
def my_replacement(match_obj):
    # 1. Достаем найденный текст
    content = match_obj.group()
    # 2. Делаем какую-то логику (например, upper())
    new_content = content.upper()
    # 3. ОБЯЗАТЕЛЬНО возвращаем строку
    return new_content
```

### Пример 1. Крик в чате

Давайте найдем все слова длиной от 5 букв и более и переведем в верхний регистр.

```python
import re


def make_loud(match):
    word = match.group()
    return word.upper()


text = 'hello my dear friend'
result = re.sub(r'\w{5,}', make_loud, text)
print(result)
```

### Пример 2. Математика в строке

Это классика. У нас есть список товаров, нужно умножать все цены на 2

```python
import re


def double_price(match):
    price_str = match.group()
    price_int = int(price_str)
    new_price = price_int * 2

    return str(new_price)


text = 'Coffee: 100$, Cake: 250$'
result = re.sub(r'\d+', double_price, text)
print(result)
```

## re.finditer()

Мы изучили <b>re.findall()</b>, который возвращает список всех совпадений. Это удобно, но у этого метода есть два недостатка:
1. <b>Память</b>: Если вы обрабатываете книгу "Война и Мир" или лог-файл на 10 Гигабайт, findall попытается найти ВСЕ совпадения сразу и создать огромный список в оперативной памяти. Ваш компьютер может "зависнуть".
2. <b>Потеря координат</b>: findall возвращает просто текст (строки). Он не говорит, где (на какой строке или позиции) этот текст был найден.

Для профессиональных задач используется <b>re.finditer()</b>

### Ленивый итератор

Название говорит само за себя: <b>Find Iterator</b>.

Эта функция не делает всю работу сразу. Она создает итератор - объект, который выдает совпадения по одному, по требованию.

### Возвращает объекты, а не строки

Второе критическое отличие:

В то время как findall возвращает строки, finditer возвращает <b>Match Objects</b> (те же самые "коробки", что и re.search).

Это значит, что для каждого найденного элемента вам доступны методы:
- <b>.group()</b> - сам текст
- <b>.start()</b> - индекс начала
- <b>.end()</b> - индекс конца
- <b>.span()</b> - кортеж координат

### Пример: Подсветка синтаксиса

Представьте, что мы пишем редактор кода. Нам нужно найти все команды def (объявления функций) и узнать их координаты, чтобы подсветить их синим цветом.

findall здесь бесполезен (он вернет список ['def', 'def'], но не скажет, где они). А finditer идеален.

```python
import re

code = """
def func1():
    pass

def func2():
    return True
"""

# Ищем слово def
# Используем finditer, так как нам нужны координаты
matches = re.finditer(r"def", code)

for match in matches:
    # match — это объект!
    print(f"Найдено: {match.group()}")
    print(f"Координаты: {match.span()}")
    print("-" * 10)

# Вывод:
# Найдено: def
# Координаты: (1, 4)
# ----------
# Найдено: def
# Координаты: (18, 21)
# ----------
```

### Экономия памяти (RAM)

Так как finditer - это генератор, он не хранит в памяти все совпадения сразу. Даже если в файле миллион совпадений, в каждый конкретный момент времени в памяти находится только один объект match.

Это позволяет обрабатывать файлы любого размера.

## re.compile()

До сих пор мы использовали "быстрыея" функции модуля: re.search(), re.findall(). Каждый раз, когда вы вызываете такую функцию, Python делает невидимую работу:
1. Берет вашу строку-шаблон r'\d+'
2. Проверяет, корректа ли она.
3. <b>Компилирует</b> её во внутренний формат (байт-код на языке С), который понятен машине.
4. Только потом ищет совпадения.

Если вы делаете это один раз - проблем нет. Но если вы делаете это в цикле на миллион итераций, вы заставляете Python тратить время впустую.

Здесь на сцену выходит <b>re.compile()</b>

### Фабрика шаблонов

Функция re.compile(pattern) берет строку регулярного выражения, компилирует её один раз и возвращает Объект шаблона (Pattern Object).

У этого объекта есть те же самые методы: .search(), .findall(), .sub(), но пользоваться ими нужно немного иначе:

```python
# regex_obj = re.compile(pattern)
# regex_obj.search(text)
```

### Когда использовать compile?

Есть две главные причины.

1. <b>Производительность в циклах</b>. Если вам нужно применить один и тот же шаблон к тысячам строк, лучше скомпилировать его один раз перед циклом.

```python
import re

lines = ["row1", "row2", "row3"] * 1000  # Много строк

# ✅ ХОРОШО:
# Компилируем 1 раз
pattern = re.compile(r"\d+") 

for line in lines:
    # Используем готовый объект. Это быстрее.
    pattern.search(line) 

# ❌ МЕНЕЕ ЭФФЕКТИВНО:
for line in lines:
    # Python вынужден проверять и обрабатывать строку шаблона на каждом шаге
    re.search(r"\d+", line)
```

Справедливости ради: у модуля re есть свой внутренний кжш. Если вы используете одну и ту же строку шаблона, Python запомнит её. Но compile позволяет избежать даже обращения к этому кэшу, что дает небольшой выигрыш в скорости.

2. <b>Чистота кода и Повторное использование</b>. Это даже важнее скорости. compile позволяет вынести все ваши регулярки в начало файла (как константы). Представьте, что у вас большой скрипт, который проверяет данные пользователя.

```python
# Блок констант в начале файла
# Мы даем регуляркам понятные имена
EMAIL_REGEX = re.compile(r"[\w\.-]+@[\w\.-]+")
PHONE_REGEX = re.compile(r"\d{10}")
DATE_REGEX  = re.compile(r"\d{4}-\d{2}-\d{2}")

def validate_user(email, phone):
    if not EMAIL_REGEX.fullmatch(email):
        print("Bad email")
    if not PHONE_REGEX.fullmatch(phone):
        print("Bad phone")
```

### Задача 1. Навигатор

<b>Условие</b>:

Функция findall возвращает только текст, но иногда нам нужно знать, где именно этот текст находится.
Для этого используется re.finditer, который возвращает объекты Match. У них есть метод .span(), возвращающий кортеж (start, end).

Найдите все вхождения слова Python (с большой буквы).
Для каждого совпадения выведите строку: Found 'Python' at (start, end).

<b>Формат ввода</b>:
Строка текста.

<b>Формат вывода</b>:
Строки с координатами.

```python
import re

pattern = re.compile(r'Python')
text = input()

for match in re.finditer(pattern, text):
    print(f'Found \'Python\' at {match.span()}')
```

### Задача 2. Компиляция шаблона

<b>Условие</b>:

Представьте, что вы обрабатываете список файлов. Чтобы не заставлять Python каждый раз заново разбирать регулярное выражение, хорошей практикой считается скомпилировать его один раз.
1. Скомпилируйте шаблон r"\.jpg$" (файлы, заканчивающиеся на .jpg) в переменную pattern.
2. Разбейте входной текст на строки (имена файлов) с помощью text.splitlines().
3. Пройдитесь по списку файлов и проверьте каждый с помощью метода pattern.search(filename).
4. Выведите только те имена файлов, которые подходят под шаблон.

<b>Формат ввода</b>:
Список файлов (каждый с новой строки).

<b>Формат вывода</b>:
Список найденных .jpg файлов.

```python
import re
import sys

match_object = re.compile(r'\.jpg$')
text = sys.stdin.read().split('\n')

for file in text:
    match = match_object.search(file)
    
    if match:
        print(file)
```

### Задача 3. Сумматор чисел

<b>Условие</b>:

re.finditer полезен, когда нужно не просто найти строки, а сразу что-то с ними сделать (например, превратить в число), не сохраняя огромный список строк в память.

В тексте содержатся цены в долларах (целые числа).
Найдите все числа (шаблон \d+), преобразуйте их в int и посчитайте их сумму.

<b>Формат ввода</b>:
Текст с числами.

<b>Формат вывода</b>:
Одно число — сумма.

```python
import re

pattern = re.compile(r'\d+')
text = input()
result = sum(map(int, map(lambda x: x.group(), pattern.finditer(text))))

print(result)
```

### Задача 4. Сложные объекты

<b>Условие</b>:

У нас есть строка с параметрами в формате key=value.
Например: width=1920 height=1080.

Нам нужно распарсить это и вывести в красивом формате.
Используйте re.finditer и шаблон с двумя группами: (\w+)=(\d+).
Внутри цикла обращайтесь к match.group(1) (ключ) и match.group(2) (значение).

<b>Формат ввода</b>:
Строка с параметрами.

<b>Формат вывода</b>:
Строки вида Key: ..., Value: ...

```python
import re

pattern = re.compile(r'(\w+)=(\d+)')
text = input()

for match in pattern.finditer(text):
    print(f'Key: {match.group(1)}, Value: {match.group(2)}')
```

### Задача 5. Анализ слов

<b>Условие</b>:

Нужно найти все слова (\w+), которые длиннее 4 букв.
Для каждого такого слова выведите его само, его длину и позицию начала.

Формат вывода: Word: {слово}, Length: {длина}, Start: {индекс}.
Используйте re.finditer.

<b>Формат ввода</b>:
Текст.

<b>Формат вывода</b>:
Список описаний слов.

```python
import re

pattern = re.compile(r'\w{4,}')
text = input()

for match in pattern.finditer(text):
    print(f'Word: {match.group()}, Length: {len(match.group())}, Start: {match.start()}')
```
