## Якоря: Привязка к краям

Мы научились искать текст. Но часто там нужно знать не только <b>что</b> мы ищем, но и <b>где именно</b> это находится.

Представьте, что вы проверяете, является ли файл картинкой.
- <b>Файл</b>: virus.jpg.exe
- <b>Шаблон</b>: .jpg

Обычный поиск найдет .jpg в середине названия. Ваш скрипт решит, что это картинка, откроет её и ... заразит компьютер.

Нам нужно убедиться, что .jpg стоит <b>в самом конце</b> строки. Для этого используются <b>Якоря</b> (<b>Anchors</b>).

### Что такое якоря?

<b>Якорь</b> - это специальный метасимвол, который не соответствует ни одному реальному символу текста. Он имеет "<b>нулевую ширину</b>" и проверяет <b>позицию</b> (место в строке).

1. Символ <b>"^"</b> - начало строки.
2. Символ <b>"$"</b> - конец строки.

### Строгая проверка (^...$)

Если мы используем оба якоря одновременно, то мы требуем полного совпадения всей строки с шаблоном. Никаких лишних символов слева или справа не можеть быть.

### re.fullmatch

В re есть функция заменяющая "^...$" - re.fullmatch(pattern, string)

## Квантификаторы: Вопросительный знак ("?")

До этого момента наши шаблоны были жесткими. Если мы писали "abc", мы требовали, чтобы в тексте были именно эти три буквы.

Но в реальной жизни данные часто бывают необязательными.
- В слове "file" может быть, а может не быть буквы "s" на конце (files)
- В ссылке может быть "http", а можеть быть "https"
- В номере телефона дефис может стоять, а может отсутствовать

Чтобы описать такую гибкость, нам нужны <b>квантификаторы</b> (указатели количества).

### Что значит "?"

В регулярных выражениях знак вопроса ("?") означает: "Символ стоящий СЛЕВА от меня, может встречаться 0 или 1 раз".

<b>Пример</b>:
pattern = r'https?'

Этот шаблон найдет и "http", и "https"

## Звездочка ("*") и Плюс ("+")

Знак вопроса ? был скромным — он разрешал символу появиться всего один раз.

Но что, если мы хотим найти слово "Heeeeeeey", где буква e повторяется много раз? Или найти число, состоящее из десятка цифр? Для этого у нас есть «тяжелая артиллерия» повторений: * и +.

### Плюс: "Один и более"

Знак плюса "+" - это требовательный квантификатор.

Символ СЛЕВА должен быть: 1 раз и больше. Если символа нет совсем - совпадение не сработает.

Шаблон: a+
- a
- aaaaa...

### Звездочка: "Ноль и более"

Символ СЛЕВА можеть быть: 0 раз и более.

Шаблон: a*
- a
- aaaa....
- ...

### К чему применяются квантификаторы?

До сих пор мы ставили квантификаторы только после одиночных букв. Но их истинная сила в том, что они применяются к предыдущему логическому элементу. Этим элементом может быть не только буква, но и набор [] или специальный символ:
- \d+
- [a-z]+
- [- ]?

## Фигурные скобки

Шорткаты <b>"+"</b> и <b>"*"</b> - это прекрасно, но они не точны. <b>"\d+"</b> найдет одну и БОЛЕЕ цифр подряд. А что если есть строгий стандарт?
- ПИН-код: ровно 4 цифры
- Год рождения: ровно 4 цифры?
- и т.д

Для точного числа символов используют <b>фигурны скобки "{}"</b>:
1. <b>Точное число повторений {n}</b>. Шаблон: \d{4}
2. <b>Диапазон {min,max}</b>. Шаблон: \w{3,10}
3. <b>Открытый диапазон {n,}</b>. Шаблон: \d{2,}

## Режимы

<b>Квантификатор</b> — это символ, который говорит "сколько раз повторить предыдущий символ/группу".

Но у него есть выбор: захватить как можно больше (жадный) или захватить как можно меньше (ленивый).

|Жадный|Ленивый|Название|Значение|
|-|-|-|-|
|*|?|Звездочка|0 или более раз|
|+|+?|Плюс|1 или более раз|
|?|??|Вопрос|0 или 1 раз|
|{n,m}|{n,m}?|Фигурные скобки|от n до m раз|
|{n,}|{n,}?|От n и более|от n раз|

Главное правило:
- <b>Жадный</b> — захватывает <b>максимально длинную</b> строку, которая подходит под шаблон.
- <b>Ленивый</b> — захватывает <b>минимально короткую</b> строку, которая подходит под шаблон.

### Задача 5. Демонстрация Жадности

<b>Условие</b>:

Это задача на понимание того, как работает движок.
У нас есть строка, состоящая из 10 букв A: AAAAAAAAAA.

Мы применяем к ней шаблон A{3,7} (от 3 до 7 букв).
Поскольку регулярные выражения жадные, движок попытается взять максимум (7), а не минимум (3).

Ваша задача:
1. Напишите шаблон A{3,7}.
2. Найдите совпадение в строке с помощью re.match.
3. Выведите длину найденной строки (len(match.group())).
Убедитесь, что программа выведет 7, а не 3.

<b>Формат ввода</b>:
Строка из букв A (например, AAAAAAAAAA).

Число (длина совпадения).
