# Области видимости переменных в Python

## Что такое область видимости в Python

<b>Область видимости</b> - это часть программы, в которой можно обратиться к переменной по её имени. Место, где переменной впервые присваивается значение, определяет её область видимости.

```python
x = 10

def my_func():
    # Переменная, созданная внутри функции, доступна только здесь
    y = 5
    print(x)  # можно обратиться к глобальной x
    print(y)  # можно обратиться к локальной y

my_func()
print(y)  # Ошибка! y не доступна вне функции
```

## Правило LEGB в Python

Python ищет переменную по порядку: <b>L -> E -> G -> B</b>. Как только имя найдено, поиск прекращается.

| Уровень | Название | Описание |
|-|-|-|
| L | Local | Локальная область внутри функции |
| E | Enclosing | Внешние (объёмлющие) функции |
| G | Global | Глобальная область модуля |
| B | Built-in | Встроенные имена Python |

## Локальные переменные (Local)

Переменные, созданные внутри функции, являются <b>локальными</b>. Они доступны только внутри этой функции.

```python
def func():
    local_var = 10   # локальная переменная
    print(local_var)

func()                # 10
print(local_var)   # Ошибка: NameError
```

Параметры функции также являются локальными переменными.

```python
def greet(name):     # name — локальная переменная
    print(f"Привет, {name}")

greet("Анна")
print(name)        # Ошибка: name не существует вне функции
```

## Внешние переменные (Enclosing)

Переменные из объёмлющей (внешней) функции доступны во вложенной функции.

```python
def outer():
    outer_var = "я снаружи"   # переменная внешней функции
    
    def inner():
        print(outer_var)       # доступ к переменной внешней функции
    
    inner()
    
outer()   # 'я снаружи'
```

## Глобальные переменные (Global)

Переменные, объявленные на верхнем уровне модуля (вне всех функций), являются глобальными. Они доступны в любой части кода.

```python
global_var = 100   # глобальная переменная

def show():
    print(global_var)   # доступ к глобальной переменной
    
show()   # 100
print(global_var)   # 100 — доступна и вне функции
```

## Встроенные переменные (Built-in)

В Python есть встроенные функции и имена, такие как print(), len(), range(). Они доступны всегда.

```python
# Встроенная функция print доступна везде
print("Это встроенная функция")
# Можно посмотреть все встроенные имена
print(dir(__builtins__))
```

Если создать переменную с таким же именем, она перекроет встроенную.

```python
# print = 5          # создали глобальную переменную print
# print(10)          # Ошибка! print теперь число, а не функция
```

## Оператор global в Python

По умолчанию внутри функции присваивание создает локальную переменную. Чтобы изменить глобальную переменную, используйте global.

```python
counter = 0        # глобальная переменная

def increment():
    global counter   # говорим, что используем глобальную
    counter += 1
    
increment()
increment()
print(counter)      # 2
```

<b>Важно</b>: без global внутри функции создастся новая локальная переменная.

```python
x = 10

def change():
    x = 20          # создаётся ЛОКАЛЬНАЯ x
    print(x)        # 20
    
change()
print(x)            # 10 (глобальная не изменилась)
```

## Оператор nonlocal для вложенных функций

<b>nonlocal</b> позволяет изменять переменные из <b>внешней (объёмлющей)</b> функции внутри вложенной.

```python
def outer():
    count = 0
    
    def inner():
        nonlocal count   # берём переменную из outer, не из global
        count += 1
        print(f"Счётчик: {count}")
    
    inner()
    inner()
    
outer()
```

Без nonlocal внутри inner() создалась бы локальная переменная, а count из outer() не изменилась.

## Пространство имен в Python

<b>Пространство имен (namespace)</b> - это словарь, который хранит имена переменных и их значения.

- <b>Глобальное пространство имен</b> - для переменных модуля.
- <b>Локальное пространство имен</b> - для переменных функции. 
- <b>Встроенное пространство имен</b> - для встроенных функций.

## Краткое резюме

| Уровень | Ключевое слово | Где объявляется | Где видна |
|-|-|-|-|
| Local | - | Внутри функции | Только внутри этой функции |
| Enclosing | nonlocal | Во внешней функции | Внутри вложенной функции |
| Global | global | На верхнем уровне модуля | Везде в модуле |
| Built-in | - | Встроенные в Python | Везде |

<b>Правила</b>: 

- Чтение перменной: поиск по LEGB

- Присваивание внутри функции: по умолчанию создается локальная переменная

- Чтобы изменить глобальную -> global 

- Чтобы изменить переменную внешней функции -> nonlocal
